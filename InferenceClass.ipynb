# YOLO instance segmentation for OCR

Use a YOLO instance-segmentation checkpoint with a class named **line**. Set the page directory and checkpoint before running the export.

```python
from pathlib import Path

from yolo_kit.inference import YoloInference
from yolo_kit.utils import get_directory_images, show_image

page_dir = Path("TestImages")
checkpoint = Path("Models/best.pt")
output_dir = Path("runs/yolo_line_export")

if not checkpoint.is_file():
    raise FileNotFoundError(f"Set checkpoint to a local line segmentation model. Missing {checkpoint}")
images = get_directory_images(page_dir)
if not images:
    raise ValueError(f"No supported page images found in {page_dir}")
inference = YoloInference(str(checkpoint))
```

```python
line_masks, page = inference.get_line_masks(images[0])
print(f"Detected {len(line_masks)} lines")
if line_masks:
    show_image(line_masks[0])
```

```python
inference.generate_page_xmL(str(page_dir), str(output_dir / "page"))
inference.generate_parquet_data(str(page_dir), str(output_dir / "parquet"))
```

```python
lines = inference.get_ocr_lines(images[0])
if lines:
    show_image(lines[0])
```
